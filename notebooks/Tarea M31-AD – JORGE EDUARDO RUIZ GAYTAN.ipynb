## Manipulación y análisis de datos FIFA con Pandas

En este notebook se utilizará Pandas para realizar análisis estadístico,
agrupaciones, incorporación de nuevos registros y creación de nuevas
variables mediante funciones lambda.

```python
import pandas as pd

ruta = r"C:\Users\jorge\OneDrive\Escritorio\TAREAM28\fifa_eda.csv"

df_fifa = pd.read_csv(ruta)
```

```python
estadisticas_generales = df_fifa[
    ["Age", "Overall", "Potential"]
].agg(["mean", "min", "max"])

estadisticas_generales
```

```python
estadisticas_pais = (
    df_fifa
    .groupby("Nationality")[["Age", "Overall", "Potential"]]
    .agg(["mean", "min", "max"])
)

estadisticas_pais
```

```python
jugador1["ID"] = 999001
jugador1["Name"] = "Jorge Ruiz"
jugador1["Age"] = 25
jugador1["Nationality"] = "Mexico"
jugador1["Overall"] = 88
jugador1["Potential"] = 90
jugador1["Club"] = "FC Data Analytics"

jugador2["ID"] = 999002
jugador2["Name"] = "Alex Torres"
jugador2["Age"] = 22
jugador2["Nationality"] = "Mexico"
jugador2["Overall"] = 82
jugador2["Potential"] = 84
jugador2["Club"] = "Python United"
```

### Comparación

Después de añadir los dos jugadores ficticios, se calcularon nuevamente
la media, el mínimo y el máximo de las variables Age, Overall y Potential.

La media puede presentar pequeñas variaciones debido a la incorporación
de los nuevos registros, mientras que los valores mínimos y máximos
solamente cambiarán si alguno de los nuevos jugadores supera los límites
existentes en el dataset original.

```python
info_paises = {
    "Mexico": ("Mexico City", "UTC-6"),
    "Argentina": ("Buenos Aires", "UTC-3"),
    "Brazil": ("Brasilia", "UTC-3"),
    "Spain": ("Madrid", "UTC+1"),
    "France": ("Paris", "UTC+1"),
    "Germany": ("Berlin", "UTC+1"),
    "England": ("London", "UTC+0"),
    "Portugal": ("Lisbon", "UTC+0"),
    "Italy": ("Rome", "UTC+1"),
    "Belgium": ("Brussels", "UTC+1")
}
```

```python
df_fifa["Capital"] = df_fifa["Nationality"].apply(
    lambda pais: info_paises.get(pais, ("No disponible", "No disponible"))[0]
)

df_fifa["Zona_Horaria"] = df_fifa["Nationality"].apply(
    lambda pais: info_paises.get(pais, ("No disponible", "No disponible"))[1]
)
```

```python
df_fifa[
    ["Name", "Nationality", "Capital", "Zona_Horaria"]
].head(20)
```

```python
df_fifa["Candidato"] = df_fifa["Potential"].apply(
    lambda potencial: "Candidato" if potencial > 85 else "Next Year"
)
```

```python
df_fifa[
    ["Name", "Overall", "Potential", "Candidato"]
].head(20)
```

```python
candidatos = df_fifa[
    df_fifa["Candidato"] == "Candidato"
]

candidatos[
    ["Name", "Nationality", "Overall", "Potential", "Candidato"]
]
```
