## Manejo de excepciones y debugging en Python

En este notebook se presentan ejemplos de manejo de excepciones utilizando
try, except, else y finally, además de un ejercicio básico de debugging.

## Ejercicio 1: Intento de abrir un archivo inexistente

En este ejercicio se intenta abrir el archivo `test.txt`.
El archivo no existe en el directorio actual, por lo que Python generará
una excepción de tipo `FileNotFoundError`.

```python
archivo = open("test.txt", "r")
contenido = archivo.read()
archivo.close()
```

## Ejercicio 2: Manejo de la excepción con TRY/EXCEPT

Ahora se utiliza una estructura `try/except` para capturar el error
`FileNotFoundError` y evitar que el programa termine inesperadamente.

```python
try:
    archivo = open("test.txt", "r")
    contenido = archivo.read()
    archivo.close()

except FileNotFoundError:
    print("Error: El archivo 'test.txt' no existe en el directorio.")
```

## Ejercicio 3: Uso de TRY/EXCEPT/ELSE/FINALLY

En este ejemplo se amplía el manejo de excepciones utilizando `else`
y `finally`.

- `try`: contiene el código que puede generar una excepción.
- `except`: captura y maneja el error.
- `else`: se ejecuta únicamente si no ocurre una excepción.
- `finally`: se ejecuta siempre, exista o no una excepción.

```python
try:
    archivo = open("test.txt", "r")
    contenido = archivo.read()

except FileNotFoundError:
    print("Error: No se encontró el archivo 'test.txt'.")

else:
    print("El archivo se abrió correctamente.")
    print(contenido)
    archivo.close()

finally:
    print("El proceso de búsqueda del archivo ha finalizado.")
```

## Ejercicio 4: Debugging de una operación matemática

Se realizará la suma de cuatro números y posteriormente se dividirá
el resultado entre cero de manera intencional.

El objetivo es identificar el error, analizar su causa y corregir el código.

```python
numero1 = 10
numero2 = 20
numero3 = 30
numero4 = 40

suma = numero1 + numero2 + numero3 + numero4

print("Resultado de la suma:", suma)

resultado = suma / 0

print("Resultado de la división:", resultado)
```

### Pasos realizados durante el debugging

1. Se ejecutó el código original.
2. Se verificó que la suma de los cuatro números se realizara correctamente.
3. Python mostró una excepción `ZeroDivisionError`.
4. Se identificó que el problema se encontraba en la operación `suma / 0`.
5. Se determinó que una división entre cero no es una operación matemática válida.
6. Se modificó el divisor por un valor diferente de cero.
7. Se volvió a ejecutar el código para comprobar que funcionara correctamente.

```python
numero1 = 10
numero2 = 20
numero3 = 30
numero4 = 40

suma = numero1 + numero2 + numero3 + numero4

print("Resultado de la suma:", suma)

divisor = 2

resultado = suma / divisor

print("Resultado de la división:", resultado)
```
