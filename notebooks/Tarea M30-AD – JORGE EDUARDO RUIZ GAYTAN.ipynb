```python
def convertir_dinero(valor):
    if pd.isna(valor):
        return None
    
    valor = str(valor).replace("€", "")
    
    if "M" in valor:
        return float(valor.replace("M", "")) * 1000000
    
    elif "K" in valor:
        return float(valor.replace("K", "")) * 1000
    
    else:
        return float(valor)
```

```python
df_fifa["Wage_numeric"] = df_fifa["Wage"].apply(convertir_dinero)

df_fifa["ReleaseClause_numeric"] = df_fifa["Release Clause"].apply(convertir_dinero)
```

```python
jugadores_por_anio = (
    df_fifa
    .groupby("Joined")
    .size()
    .reset_index(name="Numero_Jugadores")
)

jugadores_por_anio
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    jugadores_por_anio["Joined"],
    jugadores_por_anio["Numero_Jugadores"]
)

plt.title("Número de jugadores por año de incorporación")
plt.xlabel("Año de incorporación")
plt.ylabel("Número de jugadores")
plt.xticks(rotation=45)

plt.show()
```
